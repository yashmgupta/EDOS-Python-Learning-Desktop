# Chapter 7 - CSV Biology Data Analysis
This notebook combines the whole course: functions, CSV data, pandas, filtering, grouping, and plotting.

The dataset is synthetic and is included with the curriculum as `datasets/dna_samples.csv`.


## 1. Load pandas and the CSV
If pandas is not ready in EDOS, use the Package Manager or `%pip install pandas` first.


```python
import pandas as pd
import matplotlib.pyplot as plt

path = 'datasets/dna_samples.csv'
df = pd.read_csv(path)
df
```

## 2. Inspect the table


```python
print(df.columns.tolist())
print('Rows:', len(df))
df.head()
```

## 3. Reuse a GC function


```python
def gc_percent(dna):
    dna = dna.upper()
    return (dna.count('G') + dna.count('C')) / len(dna) * 100

df['length'] = df['sequence'].str.len()
df['gc_percent'] = df['sequence'].apply(gc_percent)
df[['sample_id', 'species', 'length', 'gc_percent']]
```

## 4. Find the highest-GC sample


```python
highest = df.loc[df['gc_percent'].idxmax()]
print(highest[['sample_id', 'species', 'gc_percent']])
```

## 5. Filter samples
Keep samples with GC percentage at least 50.


```python
high_gc = df[df['gc_percent'] >= 50]
high_gc[['sample_id', 'species', 'gc_percent']]
```

## 6. Group by species


```python
species_summary = df.groupby('species').agg(
    mean_gc=('gc_percent', 'mean'),
    mean_length=('length', 'mean'),
    samples=('sample_id', 'count')
)
species_summary
```

## 7. Plot GC percentage
In EDOS, the figure should appear directly below this Jupyter cell.


```python
plt.figure(figsize=(7, 4))
plt.bar(df['sample_id'], df['gc_percent'])
plt.axhline(50, linestyle='--')
plt.title('GC percentage by DNA sample')
plt.xlabel('Sample')
plt.ylabel('GC %')
plt.show()
```

## 8. Optional second dataset: gene expression
The curriculum also includes `datasets/gene_expression.csv`. Let's calculate fold change very simply.


```python
expr = pd.read_csv('datasets/gene_expression.csv')
expr['fold_change'] = expr['treated'] / expr['control']
expr.sort_values('fold_change', ascending=False)
```
