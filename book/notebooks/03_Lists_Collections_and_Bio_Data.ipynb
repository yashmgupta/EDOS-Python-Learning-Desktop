# Chapter 3 - Lists and Collections
Biology usually has many samples. Collections help us organize them.


```python
samples = ['control_1', 'control_2', 'treated_1']
print(samples)
print('How many?', len(samples))
print('First:', samples[0])
```

## List methods


```python
genes = ['TP53', 'BRCA1', 'MYC']
genes.append('EGFR')
genes.remove('MYC')
genes.sort()
print(genes)
```

## Numeric lists


```python
expression = [12.4, 20.1, 8.2, 15.7]
print('Minimum:', min(expression))
print('Maximum:', max(expression))
print('Total:', sum(expression))
print('Mean:', sum(expression) / len(expression))
```

## Dictionary for one biological sample


```python
sample = {
    'id': 'S01',
    'species': 'E. coli',
    'sequence': 'ATGCGT'
}
print(sample['species'])
sample['condition'] = 'control'
print(sample)
```

## Sets keep unique values


```python
dna = 'AAGGCTTTCCAA'
print(set(dna))
```
