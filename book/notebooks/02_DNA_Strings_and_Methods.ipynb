# Chapter 2 - DNA Strings and Methods
We will treat DNA as a Python string and explore positions, slices, and methods.


```python
dna = 'ATGCGTACGTTAGC'
print(dna)
print('Length:', len(dna))
```

## Indexing
Python starts counting at zero.


```python
print('First base:', dna[0])
print('Second base:', dna[1])
print('Last base:', dna[-1])
```

## Slicing
`dna[start:stop]` includes `start` and stops before `stop`.


```python
print('First codon:', dna[0:3])
print('Bases 4 to 9:', dna[3:9])
```

## Useful string methods


```python
dna2 = ' atgcgtaacgtt '
clean = dna2.strip().upper()
print(clean)
print('A:', clean.count('A'))
print('CG position:', clean.find('CG'))
print('Replace T with U:', clean.replace('T', 'U'))
```

## Calculate GC percentage


```python
dna = 'ATGCGCGTAA'
gc_count = dna.count('G') + dna.count('C')
gc_percent = gc_count / len(dna) * 100
print('GC %:', gc_percent)
```
