# Chapter 6 - Classes and DNA Objects
A class is a blueprint. An object is one sample made from that blueprint.


## First DNA sample class


```python
class DNASample:
    def __init__(self, sample_id, sequence):
        self.sample_id = sample_id
        self.sequence = sequence.upper()

s1 = DNASample('S01', 'atgcgt')
print(s1.sample_id)
print(s1.sequence)
```

## Add methods


```python
class DNASample:
    def __init__(self, sample_id, species, sequence):
        self.sample_id = sample_id
        self.species = species
        self.sequence = sequence.upper()

    def length(self):
        return len(self.sequence)

    def gc_percent(self):
        gc = self.sequence.count('G') + self.sequence.count('C')
        return gc / len(self.sequence) * 100

    def summary(self):
        return f'{self.sample_id} | {self.species} | {self.length()} bases'
```

```python
samples = [
    DNASample('S01', 'E. coli', 'ATGCGT'),
    DNASample('S02', 'Yeast', 'GGGCCC'),
    DNASample('S03', 'Arabidopsis', 'ATATAT')
]
for sample in samples:
    print(sample.summary(), '| GC:', round(sample.gc_percent(), 1))
```
