## Variables are like labeled sample tubes
The variable name is the label. The value is what is inside.


```python
organism = 'E. coli'
colonies = 24
temperature = 37.0
is_control = True

print(organism)
print(colonies)
print(temperature)
print(is_control)
```

## Check the type
Python can tell you what kind of value it is storing.


```python
print(type(organism))
print(type(colonies))
print(type(temperature))
print(type(is_control))
```

## DNA is just text to Python
A DNA sequence can be stored as a string.


```python
dna = 'AGCTTTTCATTCTGAC'
print('DNA:', dna)
print('Length:', len(dna))
```
