## if / else


```python
gc_percent = 62
if gc_percent > 50:
    print('GC-rich sequence')
else:
    print('AT-rich sequence')
```

## Loop through DNA


```python
dna = 'ATGC'
for base in dna:
    print(base)
```

## Count A manually


```python
dna = 'AGCTTTTCATTCTGAC'
a_count = 0
for base in dna:
    if base == 'A':
        a_count += 1
print('A count:', a_count)
```

## Position + value with enumerate()


```python
dna = 'ATGCGT'
for position, base in enumerate(dna):
    print(position, base)
```

## Scan for ATG motifs


```python
dna = 'CCATGAAATGCCC'
motif = 'ATG'
for i in range(len(dna) - len(motif) + 1):
    if dna[i:i+len(motif)] == motif:
        print('Found ATG at position', i)
```
