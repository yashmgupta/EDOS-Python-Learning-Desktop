# Chapter 5 - Functions
Functions turn repeated code into reusable tools.


## A function with no parameters


```python
def say_hello():
    print('Hello from the biology lab!')

say_hello()
```

## A function with a parameter


```python
def describe_sequence(dna):
    print('Sequence:', dna)
    print('Length:', len(dna))

describe_sequence('ATGCGT')
```

## return gives a result back


```python
def gc_percent(dna):
    dna = dna.upper()
    gc = dna.count('G') + dna.count('C')
    return gc / len(dna) * 100

result = gc_percent('ATGCGCGTAA')
print(result)
```

## Build a small DNA toolkit


```python
def clean_dna(dna):
    return dna.strip().upper()

def base_counts(dna):
    dna = clean_dna(dna)
    return {base: dna.count(base) for base in 'ACGT'}

print(base_counts(' atgcaa '))
```

## Use one function on several sequences


```python
sequences = ['ATGC', 'GGGCCC', 'ATATAT']
for seq in sequences:
    print(seq, gc_percent(seq))
```
